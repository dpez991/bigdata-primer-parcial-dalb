# Ejercicio 16: DataFrame de estudiantes

**Enunciado:** crear un DataFrame con información de estudiantes que incluya nombre, carrera, edad, promedio y ciudad. Mostrar las primeras 5 filas, la información general del DataFrame, calcular el promedio general de calificaciones, filtrar estudiantes con promedio mayor o igual a 80 y ordenar los registros de mayor a menor según el promedio.

**Lógica:** se utiliza la biblioteca Pandas para almacenar la información en un DataFrame. Posteriormente se aplican diferentes métodos para visualizar los datos, obtener estadísticas básicas, filtrar registros que cumplen una condición y ordenar la información según el promedio obtenido por cada estudiante.

```python
import pandas as pd

datos_estudiantes = {
    "Nombre": [
        "David", "Valeria", "Kevin", "Melissa", "Jorge",
        "Natalia", "Fernando", "Karla", "Samuel", "Daniela"
    ],
    "Carrera": [
        "Ingeniería en Sistemas", "Medicina", "Administración", "Derecho",
        "Arquitectura", "Psicología", "Contaduría", "Marketing",
        "Informática", "Diseño Gráfico"
    ],
    "Edad": [19, 19, 21, 23, 24, 19, 22, 21, 20, 23],
    "Promedio": [96, 88, 76, 85, 79, 91, 83, 87, 80, 89],
    "Ciudad": [
        "Tegucigalpa", "Tegucigalpa", "La Ceiba", "Comayagua",
        "Choluteca", "Danlí", "El Progreso", "Juticalpa",
        "Santa Rosa de Copán", "Puerto Cortés"
    ]
}

df = pd.DataFrame(datos_estudiantes)

print("Primeras 5 filas del DataFrame:")
print(df.head())

print("\nInformación general del DataFrame:")
df.info()

promedio_general = df["Promedio"].mean()
print(f"\nPromedio general de calificaciones: {promedio_general:.2f}")

print("\nEstudiantes con promedio mayor o igual a 80:")
print(df[df["Promedio"] >= 80])

print("\nEstudiantes ordenados por promedio de mayor a menor:")
print(df.sort_values(by="Promedio", ascending=False))
```

## Explicación de los métodos utilizados

- **pd.DataFrame()**: crea un DataFrame a partir de los datos proporcionados en forma de diccionario.

- **head()**: muestra las primeras filas del DataFrame. Por defecto presenta los primeros 5 registros.

- **info()**: proporciona información general del DataFrame, como cantidad de registros, columnas y tipos de datos.

- **mean()**: calcula el promedio de los valores numéricos de una columna.

- **sort_values()**: permite ordenar los registros de acuerdo con una columna específica. En este caso se utiliza para ordenar por promedio de mayor a menor.

- **df[df["Promedio"] >= 80]**: aplica un filtro para mostrar únicamente los estudiantes cuyo promedio es igual o superior a 80 puntos.

# Ejercicio 17: Análisis de ventas con pandas

**Enunciado:** crear un DataFrame de ventas que incluya fecha, vendedor, producto, categoría, cantidad y precio unitario. Calcular una nueva columna llamada TotalVenta, obtener el total vendido por vendedor, identificar el producto más vendido, determinar la categoría con mayor ingreso y ordenar las ventas de mayor a menor según el total vendido.

**Lógica:** se utiliza un DataFrame para almacenar los registros de ventas. A partir de los datos se genera una nueva columna con el valor total de cada venta multiplicando la cantidad por el precio unitario. Posteriormente se realizan agrupaciones y cálculos para obtener estadísticas de ventas, identificar los productos y categorías más destacados y ordenar la información para facilitar su análisis.

```python
import pandas as pd

ventas = {
    "Fecha": [
        "2026-05-01", "2026-05-02", "2026-05-03", "2026-05-04", "2026-05-05",
        "2026-05-06", "2026-05-07", "2026-05-08", "2026-05-09", "2026-05-10",
        "2026-05-11", "2026-05-12", "2026-05-13", "2026-05-14", "2026-05-15"
    ],
    "Vendedor": [
        "Luis", "Ana", "Carlos", "Luis", "Ana",
        "Carlos", "Luis", "Ana", "Carlos", "Luis",
        "Ana", "Carlos", "Luis", "Ana", "Carlos"
    ],
    "Producto": [
        "Laptop", "Mouse", "Teclado", "Monitor", "Laptop",
        "Impresora", "Mouse", "Monitor", "Laptop", "Teclado",
        "Mouse", "Impresora", "Monitor", "Laptop", "Teclado"
    ],
    "Categoria": [
        "Tecnología", "Accesorios", "Accesorios", "Tecnología", "Tecnología",
        "Oficina", "Accesorios", "Tecnología", "Tecnología", "Accesorios",
        "Accesorios", "Oficina", "Tecnología", "Tecnología", "Accesorios"
    ],
    "Cantidad": [2, 5, 3, 1, 2, 1, 4, 2, 1, 2, 6, 1, 3, 1, 5],
    "PrecioUnitario": [
        18000, 250, 600, 4500, 18000,
        3500, 250, 4500, 18000, 600,
        250, 3500, 4500, 18000, 600
    ]
}

df = pd.DataFrame(ventas)

df["TotalVenta"] = df["Cantidad"] * df["PrecioUnitario"]

print("DataFrame de ventas:")
print(df)

print("\nTotal vendido por vendedor:")
print(df.groupby("Vendedor")["TotalVenta"].sum())

producto_mas_vendido = df.groupby("Producto")["Cantidad"].sum().idxmax()

print("\nProducto más vendido:")
print(producto_mas_vendido)

ingreso_categoria = df.groupby("Categoria")["TotalVenta"].sum()
categoria_mayor_ingreso = ingreso_categoria.idxmax()

print("\nCategoría con mayor ingreso:")
print(categoria_mayor_ingreso)

print("\nVentas ordenadas de mayor a menor:")
print(df.sort_values(by="TotalVenta", ascending=False))
```

## Explicación de los métodos utilizados

- **pd.DataFrame()**: crea un DataFrame a partir de los datos proporcionados.

- **groupby()**: agrupa los registros según una columna específica para realizar cálculos sobre cada grupo.

- **sum()**: suma los valores de una columna numérica.

- **idxmax()**: devuelve el índice o valor asociado al dato más alto encontrado en una serie.

- **sort_values()**: ordena los registros utilizando una columna determinada. En este ejercicio se utiliza para ordenar las ventas de mayor a menor.

- **df["Cantidad"] * df["PrecioUnitario"]**: realiza una operación entre columnas para calcular el total generado por cada venta.

# Ejercicio 18: Limpieza de datos con pandas

**Enunciado:** crear un DataFrame con información incompleta de clientes que incluya nombre, edad, correo, ciudad y teléfono. Identificar los valores nulos, contar los valores nulos por columna, reemplazar ciudades vacías por "No especificada", eliminar registros que no tengan correo y mostrar el DataFrame limpio.

**Lógica:** se utiliza un DataFrame con algunos datos faltantes para simular problemas comunes en bases de datos reales. Posteriormente se emplean métodos de Pandas para detectar valores nulos, contabilizarlos por columna, reemplazar información faltante en la ciudad y eliminar registros que no contienen correo electrónico. Finalmente se muestra la información ya corregida.

```python
import pandas as pd

clientes = {
    "Nombre": [
        "David", "Valeria", "Kevin", "Melissa", "Jorge",
        "Natalia", "Fernando", "Karla"
    ],
    "Edad": [19, 19, 21, 23, 24, 19, 22, 21],
    "Correo": [
        "david@gmail.com",
        "vmarriaga@gmail.com",
        "kevin@gmail.com",
        "melissa@gmail.com",
        None,
        "natalia@gmail.com",
        "fernando@gmail.com",
        "karla@gmail.com"
    ],
    "Ciudad": [
        "Tegucigalpa",
        "Tegucigalpa",
        "La Ceiba",
        None,
        None,
        "Danlí",
        None,
        "San Pedro Sula"
    ],
    "Telefono": [
        "9999-1111",
        "8888-2222",
        None,
        "7777-3333",
        "6666-4444",
        None,
        "5555-5555",
        "4444-6666"
    ]
}

df = pd.DataFrame(clientes)

print("DataFrame original:")
print(df)

print("\nValores nulos:")
print(df.isnull())

print("\nCantidad de valores nulos por columna:")
print(df.isnull().sum())

df["Ciudad"] = df["Ciudad"].replace("", "No especificada")

df_limpio = df.dropna(subset=["Correo"])

print("\nDataFrame limpio:")
print(df_limpio)
```

## Explicación de los métodos utilizados

- **pd.DataFrame()**: crea un DataFrame a partir de los datos proporcionados.

- **isnull()**: identifica los valores nulos dentro del DataFrame y devuelve `True` donde existe un dato faltante.

- **sum()**: al utilizarse junto con `isnull()`, permite contar cuántos valores nulos existen en cada columna.

- **replace()**: reemplaza un valor por otro. En este ejercicio se utiliza para sustituir ciudades vacías por el texto "No especificada".

- **dropna()**: elimina filas que contienen valores nulos. Se utiliza con el parámetro `subset` para eliminar únicamente los registros que no tienen correo electrónico.

- **subset=["Correo"]**: indica que la validación de valores nulos debe realizarse solamente sobre la columna de correo.

# Ejercicio 19: Análisis de préstamos

**Enunciado:** crear un DataFrame con solicitudes de préstamo que incluya cliente, monto solicitado, plazo, ingreso mensual y estado de solicitud. Calcular el monto promedio solicitado, contar las solicitudes por estado, mostrar únicamente los préstamos aprobados, identificar los clientes que solicitaron más de 50,000 y agrupar las solicitudes por estado para obtener el promedio del monto solicitado.

**Lógica:** se utiliza un DataFrame para almacenar la información de las solicitudes de préstamo. Posteriormente se aplican diferentes métodos de Pandas para calcular estadísticas, filtrar registros según determinadas condiciones, contar solicitudes por estado y agrupar datos para obtener promedios según la categoría correspondiente.

```python
import pandas as pd

prestamos = {
    "Cliente": [
        "David", "Valeria", "Kevin", "Melissa", "Jorge",
        "Natalia", "Fernando", "Karla", "Samuel", "Daniela"
    ],
    "MontoSolicitado": [
        45000, 70000, 30000, 85000, 55000,
        40000, 95000, 60000, 25000, 75000
    ],
    "PlazoMeses": [
        24, 36, 12, 48, 36,
        24, 60, 36, 12, 48
    ],
    "IngresoMensual": [
        18000, 30000, 15000, 35000, 22000,
        17000, 40000, 25000, 14000, 32000
    ],
    "Estado": [
        "Aprobado", "Aprobado", "Pendiente", "Aprobado", "Rechazado",
        "Pendiente", "Aprobado", "Rechazado", "Pendiente", "Aprobado"
    ]
}

df = pd.DataFrame(prestamos)

print("Solicitudes de préstamo:")
print(df)

promedio_monto = df["MontoSolicitado"].mean()

print("\nMonto promedio solicitado:")
print(f"L. {promedio_monto:.2f}")

print("\nCantidad de solicitudes por estado:")
print(df["Estado"].value_counts())

print("\nPréstamos aprobados:")
print(df[df["Estado"] == "Aprobado"])

print("\nClientes con monto solicitado mayor a 50,000:")
print(df[df["MontoSolicitado"] > 50000])

print("\nPromedio del monto solicitado por estado:")
print(df.groupby("Estado")["MontoSolicitado"].mean())
```

## Explicación de los métodos utilizados

- **pd.DataFrame()**: crea un DataFrame a partir de los datos proporcionados.

- **mean()**: calcula el promedio de los valores numéricos de una columna.

- **value_counts()**: cuenta cuántas veces aparece cada valor dentro de una columna. En este ejercicio se utiliza para contar solicitudes según su estado.

- **df[df["Estado"] == "Aprobado"]**: aplica un filtro para mostrar únicamente los préstamos que fueron aprobados.

- **df[df["MontoSolicitado"] > 50000]**: permite mostrar solamente los clientes cuyo monto solicitado supera los 50,000.

- **groupby()**: agrupa los registros según una columna específica para realizar operaciones sobre cada grupo.

- **groupby("Estado")["MontoSolicitado"].mean()**: calcula el promedio del monto solicitado para cada estado de solicitud.

# Ejercicio 20: Mini análisis de datos tipo empresa

**Enunciado:** crear un DataFrame con información de empleados que incluya nombre, departamento, cargo, salario, años de experiencia y ciudad. Mostrar los primeros registros, obtener estadísticas generales de salarios, calcular el salario promedio por departamento, identificar el empleado con el salario más alto, mostrar empleados con más de 5 años de experiencia, ordenar los empleados por salario y crear una columna llamada NivelExperiencia.

**Lógica:** se utiliza un DataFrame para almacenar información de empleados de una empresa. A partir de estos datos se aplican métodos de Pandas para obtener estadísticas salariales, realizar agrupaciones por departamento, filtrar empleados según su experiencia y ordenar los registros. Además, se crea una nueva columna para clasificar a cada empleado como Junior, Intermedio o Senior según sus años de experiencia.

```python
import pandas as pd

empleados = {
    "Nombre": [
        "David", "Valeria", "Kevin", "Melissa", "Jorge",
        "Natalia", "Fernando", "Karla", "Samuel", "Daniela"
    ],
    "Departamento": [
        "TI", "Recursos Humanos", "Finanzas", "Marketing", "TI",
        "Ventas", "Finanzas", "Marketing", "Ventas", "TI"
    ],
    "Cargo": [
        "Desarrollador", "Analista RRHH", "Contador", "Diseñadora",
        "Administrador de Redes", "Ejecutiva de Ventas",
        "Auditor", "Community Manager", "Supervisor de Ventas",
        "Ingeniera de Software"
    ],
    "Salario": [
        28000, 22000, 26000, 24000, 32000,
        21000, 29000, 23000, 31000, 35000
    ],
    "AniosExperiencia": [
        4, 2, 6, 3, 8,
        1, 7, 5, 9, 10
    ],
    "Ciudad": [
        "Tegucigalpa", "Tegucigalpa", "La Ceiba", "Comayagua",
        "Choluteca", "Danlí", "El Progreso",
        "Juticalpa", "Santa Rosa de Copán", "Puerto Cortés"
    ]
}

df = pd.DataFrame(empleados)

def clasificar_experiencia(anios):
    if anios <= 2:
        return "Junior"
    elif anios <= 5:
        return "Intermedio"
    else:
        return "Senior"

df["NivelExperiencia"] = df["AniosExperiencia"].apply(clasificar_experiencia)

print("Primeros registros:")
print(df.head())

print("\nEstadísticas generales de salarios:")
print(df["Salario"].describe())

print("\nSalario promedio por departamento:")
print(df.groupby("Departamento")["Salario"].mean())

empleado_salario_alto = df.loc[df["Salario"].idxmax()]

print("\nEmpleado con salario más alto:")
print(empleado_salario_alto)

print("\nEmpleados con más de 5 años de experiencia:")
print(df[df["AniosExperiencia"] > 5])

print("\nEmpleados ordenados por salario:")
print(df.sort_values(by="Salario", ascending=False))
```

## Explicación de los métodos utilizados

- **pd.DataFrame()**: crea un DataFrame a partir de los datos proporcionados.

- **head()**: muestra los primeros registros del DataFrame.

- **describe()**: genera estadísticas generales de una columna numérica, como cantidad de datos, promedio, valor mínimo, máximo y desviación estándar.

- **groupby()**: agrupa los registros según una columna específica para realizar cálculos sobre cada grupo.

- **mean()**: calcula el promedio de los valores numéricos.

- **idxmax()**: devuelve la posición del valor más alto dentro de una columna.

- **loc[]**: permite acceder a una fila específica del DataFrame utilizando su índice.

- **apply()**: ejecuta una función personalizada sobre cada valor de una columna. En este ejercicio se utiliza para clasificar el nivel de experiencia de cada empleado.

- **sort_values()**: ordena los registros según una columna determinada. En este caso se utiliza para ordenar los salarios de mayor a menor.

- **df[df["AniosExperiencia"] > 5]**: filtra los empleados que poseen más de cinco años de experiencia.
